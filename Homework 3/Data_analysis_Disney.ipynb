```python
import pandas as pd

df = pd.read_csv(r"/Users/gianluca/Library/Mobile\ Documents/com\~apple\~CloudDocs/Universität/HS\ 26/Python\ Seminar/Homework/Homework-Python/Homework\ 3/disney_plus_shows.csv".replace("\\", ""))
df.head()
```

```python
df_copy = df.copy()
movies = df_copy[df_copy["type"] == "movie"]
movies.shape
```

```python
movies = movies[["imdb_id", "title", "director", "year", "imdb_rating", "imdb_votes"]]
movies.head()
```

```python
movies = movies.drop_duplicates()
movies.shape
```

```python
movies = movies[movies["director"].notna()]
movies = movies[movies["imdb_rating"].notna()]
movies = movies[movies["imdb_votes"].notna()]
movies.shape
```

```python
movies = movies.copy()
movies["imdb_votes"] = movies["imdb_votes"].str.replace(",", "")
movies["imdb_votes"] = movies["imdb_votes"].astype(int)
movies["year"] = movies["year"].astype(int)
movies.dtypes
```

```python
movies = movies[movies["imdb_votes"] >= 150]
movies.shape
```

```python
movies = movies.copy()
movies["director"] = movies["director"].str.replace("(co-director)", "")
movies.head()
```

```python
movies["director"] = movies["director"].str.split(", ")
movies.head()
```

```python
movies = movies.explode("director")
movies.shape
```

```python
counts = movies["director"].value_counts()
counts
```

```python
top_directors = counts[counts >= 4]
movies = movies[movies["director"].isin(top_directors.index)]
movies.shape
```
